# 01. Data Cleaning

## Importing libraries and datasets

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df1 = pd.read_csv("utahjazz24-25stats.csv")
```

```python
df1.head()
```

```python
df1.info()
```

```python
df1.drop(columns=["Rk", "Awards"], inplace=True)
```

```python
df1.head()
```

```python
df2 = pd.read_csv("utahjazz24-25players.csv")
```

```python
df2.head()
```

```python
df2.drop(columns=["No", "College"], inplace=True)
```

### Merging the two datasets

```python
df = pd.merge(df2, df1, on="Player", how="left")
```

```python
df.head()
```

```python
df.rename(columns={"Ht":"Height"}, inplace=True)
```

```python
df.rename(columns={"Wt":"Weight"}, inplace=True)
```

```python
df.head()
```

## Data Cleaning

### Position column

```python
df.drop(columns=["Pos_y"], inplace=True)
df.rename(columns={"Pos_x":"Pos"}, inplace=True)
```

```python
df.head()
```

Birthdate column

```python
df["Birth_Date"] = pd.to_datetime(df["Birth_Date"])
```

```python
df.head()
```

### Example for NaN values: Here i'm inserting missing data for Jaden Springer

```python
df.loc[df['Player'] == 'Jaden Springer', 
       ['G','MP','FG','FGA','FG%','3P','3PA','3P%','2P','2PA','2P%','eFG%','FT','FTA','FT%','ORB','DRB','TRB','AST','STL','BLK','TOV','PF','PTS']] = [
    17,13.2,1.4,3.3,.411,0.4,1.7,.207,1.0,1.6,.630,.464,0.7,1.0,.706,0.8,1.2,2.0,1.4,0.7,0.2,0.7,1.5,3.8]
```

```python
df.loc[df['Player'] == 'Jaden Springer', ['Age', 'GS']] = [22, 0]
```

### Moving the "Age" column near the "Birth_Date" column

```python
col = df.pop("Age")
df.insert(5, "Age", col)
```

```python
df["Age"] = df["Age"].astype(int)
```

### Changing values type

```python
df["G"] = df["G"].astype(int)
df["GS"] = df["GS"].astype(int)
```

```python
df.head()
```

### Saving the dataset

```python
df.to_csv("UtahJazz24-25clean.csv", index=False)
```
